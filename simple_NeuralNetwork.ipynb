```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# Create a Model Class that inherits nn.Module
class Model(nn.Module):
   # Inout layer (4 features of the flower iris) -->
   # Hidden layer1 (number of neurons) -->
   # HL2 (n neurons) -->
   # Output (3 classes of iris)
   def __init__(self, input_features=4, h1=8, h2=9, out_features=3):
    super().__init__() # instantiate (δημιουργώ στιγμιότυπο) our nn.Module
    self.fc1 = nn.Linear(input_features, h1)
    self.fc2 = nn.Linear(h1, h2)
    self.out = nn.Linear(h2, out_features)

    def forward(self, x):
      x = F.relu(self.fc1(x))
      x = F.relu(self.fc2(x))
      x = self.out(x)

      return x
```

```python
# Pick a manual seed for randomization
torch.manual_seed(41)
# Create an instance of Model
model = Model()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
url = 'https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv'
my_df = pd.read_csv(url)
```

```python
my_df.tail()
```

```python
# Change last columns from strings to integers
my_df['species'] = my_df['species'].replace('setosa', 0.0)
my_df['species'] = my_df['species'].replace('versicolor', 1.0)
my_df['species'] = my_df['species'].replace('virginica', 2.0)
my_df
```

```python
# Train Test Split! Set X(feature), y(output)
X = my_df.drop('species', axis=1)
y = my_df['species']
```

```python
# Convert these to numpy arrays
X = X.values
y = y.values
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=41)
```

```python
# Convert X features to float tensors
X_train = torch.FloatTensor(X_train)   # float γιατί τα X(πέταλα κλπ) είναι δεκαδικοί
X_test = torch.FloatTensor(X_test)
```

```python
# Convert y labels to long tensors (64-bit integers)
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

```python
# Set the criterion of our model to measure the error, how far the predictions are from the data
criterion = nn.CrossEntropyLoss()
# Choose Adam Optimizer, lr= learning rate (if error doesn't go down after a bunch of iterations (epochs), lower our learning rate)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
model.parameters
```

```python
# Train our model
# epochs?(one run through all training data in our NN)
epochs = 100
losses = []
for i in range(epochs):
  # Go forward and get a prediction
  y_pred = model.forward(X_train) # Get predicted results

  # Measure the loss/error, gonna be high at first
  loss = criterion(y_pred, y_train) # predicted values vs the y_train

  # Keep track of our losses
  losses.append(loss.detach().numpy())

  # Print every 10 epochs
  if i % 10 == 0:
    print (f'Epoch: {i} and loss: {loss}')

  # Do some backpropagation: take the error rate of forward propagation and feed it back
  # through the network to fine tune the weights
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
```
